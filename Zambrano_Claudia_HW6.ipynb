### (b) Data Preparation

#### (i) Research what types of techniques are usually used

To find the best type of data imputation techniques we would usually find out why or what kind of data is missing. Since the names of the features were ommitted I will use the mean of each feature to replace the missing values, since it is the most popular practice.

```python
# Replace 'na' string representations with NaN
df = df.replace('na', np.nan)

# Identify columns to impute (all columns except the 'class' column)
feature_cols = [col for col in df.columns if col != 'class']

# Convert feature columns to numeric so we can compute the mean
df[feature_cols] = df[feature_cols].apply(pd.to_numeric)

# Fill the missing values with the mean of their respective columns
df[feature_cols] = df[feature_cols].fillna(df[feature_cols].mean())

# Display the updated dataset and verify if any missing values remain
print("Missing values remaining:", df.isna().sum().sum())
df.head()
```

#### (ii) Calculate the coefficient of variation

```python
# Standard deviation and mean for each feature
sd_vals = df[feature_cols].std()
mean_vals = df[feature_cols].mean()

# Coefficient of variation (sd / mean)
cv_vals = sd_vals / mean_vals

# DataFrame to display the metrics side-by-side
cv_df = pd.DataFrame({
    'Sample SD': sd_vals,
    'Sample Mean': mean_vals,
    'Coefficient of Variation (CV)': cv_vals
})

cv_df.head(5)
```

#### (iii) Plot a correlation matrix

```python
# Calculate the correlation matrix
correlation_matrix = df[feature_cols].corr()

# Using a heatmap because there are alot of features
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, cmap='coolwarm', xticklabels=False, yticklabels=False)
plt.title('Correlation Matrix Heatmap (All 170 Features)', fontsize=16)
plt.xlabel('Features', fontsize=12)
plt.ylabel('Features', fontsize=12)
plt.show()
```
